# 2次元イジング模型：相転移の理論と計算手法の数学的解説

このノートブックでは、2次元イジング模型における相転移現象、それを解析するためのモンテカルロ法、およびビンダー累積量の理論的背景を数式のみを用いて解説します。

---

## 1. イジング模型の定義

2次元正方格子上の各サイト $i$ に、スピン変数 $s_i \in \{+1, -1\}$ を配置します。外部磁場がない場合の系のハミルトニアン（全エネルギー） $H$ は次のように定義されます：

$$ H = -J \sum_{\langle i,j \rangle} s_i s_j $$

ここで、
- $J > 0$：強磁性相互作用定数（隣接するスピンが揃うとエネルギーが下がる）。
- $\sum_{\langle i,j \rangle}$：すべての隣接するサイトのペアについての和。

---

## 2. モンテカルロ法（メトロポリス法）によるシミュレーション

現実の物質が温度 $T$ の熱浴中にあるとき、その状態は「ボルツマン分布」に従って確率的に変動します。これをコンピューター上で再現するのがメトロポリス法です。

### 2.1 更新の手順（ステップ・バイ・ステップ）

1回の更新試行において、以下の手順を踏みます：

1.  **試行**: 格子の中からランダムに1つのスピン $s_i$ を選び、それを反転させた状態（$s_i \to -s_i$）を仮定します。
2.  **エネルギー計算**: 反転させたことによって生じるエネルギーの変化 $\Delta E = E_{\text{new}} - E_{\text{old}}$ を計算します。
3.  **判定**: 以下のルールに従って、実際に反転させるかどうかを決定します。
    -   **$\Delta E \le 0$ の場合**（エネルギーが下がる、または変わらない）：
        -   系はより安定な状態を好むため、**無条件で反転を受け入れます**。
    -   **$\Delta E > 0$ の場合**（エネルギーが上がる）：
        -   **確率 $P = e^{-\Delta E / k_B T}$** で反転を受け入れ、残りの確率で元の状態を維持します。

### 2.2 物理的な意味

-   **なぜ確率的にエネルギーが上がることを許すのか？**: 現実のミクロな世界では、熱エネルギー（熱ゆらぎ）によって、一時的に不安定な状態へ跳ね上がることがあるからです。これがなければ、系は一度安定した状態から二度と動けなくなり、相転移のようなダイナミックな変化を再現できません。
-   **温度 $T$ の役割**: 温度が高いほど $e^{-\Delta E / k_B T}$ は 1 に近づき、エネルギーが上がる変化も頻繁に起きるようになります（バラバラな状態）。逆に温度が低いとこの確率は 0 に近づき、系はエネルギーの低い揃った状態へと落ち着いていきます。

## 3. ビンダー累積量 (Binder Cumulant)

相転移点（臨界点）を特定するために、磁化 $m = \frac{1}{N} \sum s_i$ の高次モーメントを用いた**ビンダー累積量** $U_L$ を導入します。

### 3.1 定義式

$$ U_L = 1 - \frac{\langle m^4 \rangle}{3 \langle m^2 \rangle^2} $$

ここで、$\langle \cdot \rangle$ は熱力学的な平均（アンサンブル平均）を表します。

### 3.2 なぜ臨界点で値が一致するのか？

臨界点付近における物理量の振る舞いを記述する「有限サイズスケーリング仮説」に基づくと、磁化の分布関数 $P(m)$ は次のような形にスケーリングされます：

$$ P(m) = L^{\beta/\nu} \tilde{P}(m L^{\beta/\nu}) $$

ここで登場するギリシャ文字は、相転移の性質を決定づける**「臨界指数」**と呼ばれる重要な定数です：
- **$\beta$（ベータ）**: 磁化の臨界指数。温度が臨界点に近づく際の、自発磁化の消え方の鋭さを表します。
- **$\nu$（ニュー）**: 相関距離の臨界指数。スピン同士が影響を及ぼし合う距離（相関距離）が、臨界点付近でどのように増大するかを表します。

このスケーリング則を用いると、$k$ 次のモーメント $\langle m^k \rangle$ は次のように計算されます：

$$ \langle m^k \rangle = \int m^k P(m) dm \propto L^{-k\beta/\nu} $$

ビンダー累積量 $U_L$ の定義式にこれを代入すると、分子の $\langle m^4 \rangle$ と分母の $\langle m^2 \rangle^2$ の双方から、サイズ依存性を示す項 $L^{-4\beta/\nu}$ が現れます：

$$ U_L = 1 - \frac{(\text{定数}) \cdot L^{-4\beta/\nu}}{3 \cdot [(\text{定数}) \cdot L^{-2\beta/\nu}]^2} $$

これにより、分子と分母で $L$ のべき乗がちょうど打ち消し合い（キャンセルされ）、**臨界点 $T = T_c$ においては $U_L$ がシステムサイズ $L$ に依存しない一定の値をとる**ことになります。このため、異なる $L$ のグラフを描くと、臨界点において一点で交差するのです。

## 4. 厳密解との比較と結果の予測

### 4.1 オンサガーの厳密解

2次元正方格子イジング模型の無限系における転移温度 $T_c$ は、以下の関係式から導かれます：

$$ \sinh\left(\frac{2J}{k_B T_c}\right) = 1 \implies T_c = \frac{2J}{k_B \ln(1+\sqrt{2})} \approx 2.269185 J/k_B $$

### 4.2 シミュレーションによる予測結果

異なるシステムサイズ（$L=8, 16, 32$ など）で計算された $U_L$ のグラフを重ね合わせ、その交差点を読み取ることで以下の予測が得られます：

- **転移温度の予測**: $T \approx 2.27$
- **臨界点における累積量の値**: $U^* \approx 0.61$

この予測値はオンサガーの厳密解と極めて高い精度で一致しており、有限サイズスケーリングの手法が正しく機能していることが示されます。
