# 【はじめての方向け】2次元イジング模型とモンテカルロ法

このノートでは、「イジング模型って何？」「なんでモンテカルロ法で計算できるの？」という疑問を、数式と直感的なイメージを使って分かりやすく解説します。

---

## 1. イジング模型とは？（磁石のモデル）

イジング模型は、鉄などの「磁石」が、温度によって磁力を持ったり失ったりする様子を計算するための最もシンプルな模型です。

### 設定
- **スピン**: ミクロな小さな磁石のことです。上（$+1$）か下（$-1$）の2つの状態しか取りません。
- **格子**: 2次元の正方形のマス目（ジャングルジムの面のようなもの）にスピンが並んでいます。

### 物理的な「願い」：エネルギー $E$
自然界は「エネルギーを低くしたい」という性質があります。イジング模型では、**隣り合う磁石が同じ向きだとエネルギーが下がる**というルールがあります。

$$ E_{pair} = -J S_i S_j $$

- $S_i, S_j$: 隣り合う磁石の向き（$+1$ または $-1$）。
- **同じ向き** ($+1$ と $+1$, または $-1$ と $-1$) のとき： $S_i S_j = +1$ なので、エネルギーは **$-J$ (低い)**。
- **逆の向き** ($+1$ と $-1$) のとき： $S_i S_j = -1$ なので、エネルギーは **$+J$ (高い)**。

系全体のエネルギー $H$ は、すべての隣り合うペアの合計です：
$$ H = -J \sum_{\langle i,j \rangle} S_i S_j $$

---

## 2. 温度 $T$ の役割（エネルギー vs ゆらぎ）

ここで「温度」が登場します。温度はスピンを「かき乱す」役割をします。

- **温度が低い ($T \approx 0$)**: 「エネルギーを低くしたい」という願いが勝ちます。みんな同じ向きに揃い、**磁石になります**（強磁性）。
- **温度が高い ($T \to \infty$)**: 「熱ゆらぎ」が勝ち、スピンはバラバラの向きになります。全体としては磁力を打ち消し合い、**磁石ではなくなります**。

この「揃う」か「バラバラか」が切り替わる境界の温度が **転移温度 $T_c$** です。

---

## 3. モンテカルロ法：どうやってシミュレーションする？

「ある温度 $T$ でスピンがどう並ぶか」を計算するのは、組み合わせが多すぎて（$2^{L^2}$ 通り）不可能です。そこで、**「サイコロを振って状態を決める」**モンテカルロ法を使います。

### メトロポリス法のアルゴリズム（直感版）
1. 格子の中からランダムに1つのスピンを選び、**「もしひっくり返したら？」**と考えます。
2. ひっくり返して**エネルギーが下がるなら、喜んでひっくり返します**。
3. エネルギーが**上がる場合でも、たまに（確率的に）ひっくり返ります**。この確率は温度が高いほど大きくなります。

確率の数式（ボルツマン因子）：
$$ P = \exp\left( -\frac{\Delta E}{T} \right) $$
- $\Delta E$: ひっくり返した時のエネルギーの増分。
- これにより、低いエネルギーの状態になりやすいけれど、熱による「乱れ」も表現できます。

---

## 4. なぜ「交点」で温度がわかるの？（ビンダー累積量）

課題で使った **ビンダー累積量 $U_L$** は、スピンの「揃い具合のバラツキ」を見る指標です。

$$ U_L = 1 - \frac{\langle m^4 \rangle}{3 \langle m^2 \rangle^2} $$

- **バラバラな時 (高温)**: $U_L$ は $0$ に近づきます。
- **完全に揃っている時 (低温)**: $U_L$ は $2/3 \approx 0.66$ に近づきます。

面白いのは、**「ちょうど相転移が起きる瞬間のバラツキ具合は、サイズ $L$ によらず共通である」**という物理的な性質があることです。そのため、複数のサイズでグラフを描くと、転移温度の場所だけでピッタリ重なり（交わり）ます。

---

## 5. まとめ：課題の意味
- **$O(L^2)$ でスケール**: 全サイトを1回ずつチェックするので、面積（$L \times L$）に比例して時間がかかるはず、という確認です。
- **有限サイズスケーリング**: コンピュータでは無限に大きな系を扱えないため、いくつかの小さなサイズの結果を比較して、無限系の真の転移温度を予言する手法です。
