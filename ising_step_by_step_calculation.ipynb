# ビンダー累積量：計算の全ステップ解説

このノートでは、シミュレーションの中で行われている計算を、1つずつ数式と具体的な数値を使って追いかけます。

---

## ステップ 1：磁化 $m$ を測る

まず、ある瞬間のスピンの並びから「磁化」を計算します。磁化は「全スピンの平均値」です。

### 数式
$$ m = \frac{1}{N} \sum_{i=1}^{N} S_i $$

### 具体例
$2 \times 2$ ($N=4$) の系で、スピンが `[+1, +1, -1, +1]` だった場合：
$$ m = \frac{1 + 1 - 1 + 1}{4} = \frac{2}{4} = 0.5 $$

---

## ステップ 2：データをたくさん集める（サンプリング）

モンテカルロ法では、スピンを動かしながら何度も磁化を測り、リストにします。

**例：5回測った結果（サンプルデータ）**
$$ m = [0.8, -0.7, 0.9, -0.8, 0.8] $$

---

## ステップ 3：2次モーメント $\langle m^2 \rangle$ の計算

集めたデータのそれぞれを**2乗**して、その平均をとります。

### 計算プロセス
1. 各データを2乗する：
   - $0.8^2 = 0.64$
   - $(-0.7)^2 = 0.49$
   - $0.9^2 = 0.81$
   - $(-0.8)^2 = 0.64$
   - $0.8^2 = 0.64$
2. 合計して個数(5)で割る：
   $$ \langle m^2 \rangle = \frac{0.64 + 0.49 + 0.81 + 0.64 + 0.64}{5} = \frac{3.22}{5} = 0.644 $$

---

## ステップ 4：4次モーメント $\langle m^4 \rangle$ の計算

同様に、各データを**4乗**して平均をとります。

### 計算プロセス
1. 各データを4乗する：
   - $0.8^4 = 0.4096$
   - $(-0.7)^4 = 0.2401$
   - $0.9^4 = 0.6561$
   - $(-0.8)^4 = 0.4096$
   - $0.8^4 = 0.4096$
2. 合計して個数(5)で割る：
   $$ \langle m^4 \rangle = \frac{0.4096 + 0.2401 + 0.6561 + 0.4096 + 0.4096}{5} = \frac{2.125}{5} = 0.425 $$

---

## ステップ 5：ビンダー累積量 $U_L$ の算出

最後に、これまでの値を公式に代入します。

### 公式
$$ U_L = 1 - \frac{\langle m^4 \rangle}{3 \langle m^2 \rangle^2} $$

### 今回の数値での代入
1. 分母を計算： $3 \times (0.644)^2 = 3 \times 0.4147 = 1.2441$
2. 分数を計算： $0.425 / 1.2441 \approx 0.3416$
3. 1から引く： $U_L = 1 - 0.3416 = \mathbf{0.6584}$

---

## 最後に：コードで書くとこうなります

この一連の流れを Python で書くと、たったこれだけになります。

```python
import numpy as np

# ステップ2：集めたデータ
ms = np.array([0.8, -0.7, 0.9, -0.8, 0.8])

# ステップ3：2次モーメント
m2 = np.mean(ms**2)

# ステップ4：4次モーメント
m4 = np.mean(ms**4)

# ステップ5：ビンダー累積量
UL = 1 - m4 / (3 * m2**2)

print(f"磁化のデータ: {ms}")
print(f"2次モーメント <m^2>: {m2:.4f}")
print(f"4次モーメント <m^4>: {m4:.4f}")
print(f"ビンダー累積量 UL: {UL:.4f}")
```
