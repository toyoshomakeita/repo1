# モンテカルロ・ステップ (MCS) とは何か？

イジング模型の計算でよく出てくる **「1 MCS」** という言葉。これは、シミュレーションにおける **「時間の単位」** のようなものです。

このノートでは、1 MCS が具体的に何を指し、なぜそう呼ぶのかを分かりやすく解説します。

---

## 1. 直感的なイメージ：全員にチャンスを与える

例えば、$L \times L = 10 \times 10$ のクラス（100人）があるとします。
「モンテカルロ法で状態を更新する」というのは、一人を選んで「君、スピンをひっくり返す？」と聞く作業です。

- **1回の更新試行**: 誰か1人を選んで、ルールに従ってひっくり返すか決める。
- **1 MCS**: クラスの人数と同じ回数（100回）、誰かを選んで聞く作業を繰り返すこと。

> **ポイント**：1 MCS 終わったとき、平均して **全員が1回ずつ「更新のチャンス」をもらった** ことになります。これが「1ステップ」の区切りになります。

---

## 2. なぜ「1回」ではなく「全サイト分」を1ステップとするのか？

もし「1人の更新」を「1ステップ」と呼んでしまうと、系のサイズ $L$ が大きくなったときに困ります。

- $L=10$ のときの1回：全体の 1/100 が変わる可能性がある。
- $L=100$ のときの1回：全体の 1/10,000 しか変わらない。

これではサイズの違う系を比較できません。そこで、**「系のサイズ（面積 $L^2$）に比例した回数」を 1ステップと定義する** ことで、サイズに関わらず「系全体がひと通り変化するのに必要な時間」を公平に測れるようになります。

---

## 4. まとめ：1 MCS の役割

1. **物理的な時間の目安**: 「1000 MCS 動かした」といえば、系全体を 1000回 かき回したことになります。
2. **平衡状態への到達**: 初期状態から $T_c$ 付近の平衡状態に達するには、数千〜数万 MCS 程度の「時間」が必要になります。
3. **統計平均**: 磁化などを測る際も、「1 MCS ごとに測る」のが一般的です。あまりに短い間隔（1回ごと）で測っても、状態がほとんど変わっていないため、統計的に独立なデータにならないからです。
